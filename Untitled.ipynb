```python
from huggingface_hub import snapshot_download
from pathlib import Path
import json
import re
import pandas as pd

# Download metadata only using your saved Hugging Face login.
root = Path(snapshot_download(
    repo_id="touhid314/cktformer-dataset",
    repo_type="dataset",
    allow_patterns=["metadata/**"],
    local_dir="cktformer",
    token=True,
    max_workers=2
))

rules = {
    "amplifier": [
        "amplifier",
        "common emitter",
        "common source",
        "cascode",
    ],
    "oscillator": [
        "oscillator",
        "astable",
    ],
    "rectifier": [
        "rectifier",
        "half wave",
        "full wave",
    ],
    "filter": [
        "low pass",
        "high pass",
        "band pass",
        "band stop",
        "notch",
    ],
    "voltage_regulator": [
        "voltage regulator",
        "linear regulator",
        "ldo",
    ],
}


def normalize(text):
    return re.sub(r"[^a-z0-9]+", " ", str(text).lower()).strip()


rows = []
metadata_files = sorted((root / "metadata").rglob("*.json"))

print(f"\nFound {len(metadata_files):,} metadata files.")

for path in metadata_files:
    meta = json.loads(path.read_text(encoding="utf-8"))

    names = meta.get("names") or []
    if isinstance(names, str):
        names = [names]

    # Match circuit names; descriptions are included for manual review.
    text = " " + normalize(" ".join(str(name) for name in names)) + " "

    matches = [
        label
        for label, keywords in rules.items()
        if any(
            " " + normalize(keyword) + " " in text
            for keyword in keywords
        )
    ]

    if not matches:
        continue

    source = meta.get("source") or {}

    rows.append({
        "id": meta.get("id", path.stem),
        "suggested_class": matches[0] if len(matches) == 1 else "REVIEW",
        "matches": "; ".join(matches),
        "names": "; ".join(str(name) for name in names),
        "description": meta.get("desc", ""),
        "book": source.get("book", ""),
        "page": source.get("page", ""),
    })

columns = [
    "id",
    "suggested_class",
    "matches",
    "names",
    "description",
    "book",
    "page",
]

df = pd.DataFrame(rows, columns=columns)

csv_path = Path("circuit_candidates.csv")
df.to_csv(csv_path, index=False, encoding="utf-8-sig")

if not df.empty:
    print("\nCandidate counts:")
    print(df["suggested_class"].value_counts().to_string())
    print(f"\nTotal candidates: {len(df):,}")
else:
    print("\nNo matches found. Check the metadata and keywords.")

print(f"\nSaved candidate list to:\n{csv_path.resolve()}")
print("\nThese are suggested labels and need visual review.")
print("No images or SPICE netlists were downloaded.")
```
